```python
import pandas as pd

customers = pd.read_csv('/content/olist_customers_dataset.csv')
geolocation = pd.read_csv('/content/olist_geolocation_dataset.csv')
order_items = pd.read_csv('/content/olist_order_items_dataset.csv')
payments = pd.read_csv('/content/olist_order_payments_dataset.csv')
# Specifying latin-1 encoding to handle special Portuguese characters
reviews = pd.read_csv('/content/olist_order_reviews_dataset.csv', encoding='latin-1')
orders = pd.read_csv('/content/olist_orders_dataset.csv')
products = pd.read_csv('/content/olist_products_dataset.csv')
sellers = pd.read_csv('/content/olist_sellers_dataset.csv')
category_translation = pd.read_csv('/content/product_category_name_translation.csv')

print("All datasets loaded successfully!")
```

```python
datasets = {
    'customers': customers,
    'geolocation': geolocation,
    'order_items': order_items,
    'payments': payments,
    'reviews': reviews,
    'orders': orders,
    'products': products,
    'sellers': sellers,
    'category_translation': category_translation
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows x {df.shape[1]} columns")
```

```python
# Preview each dataset and its column names

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)
    print("Columns:", list(df.columns))
    display(df.head())
```

```python
# Check data quality for all datasets

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)

    print("Shape:", df.shape)
    print("Duplicate rows:", df.duplicated().sum())

    print("Missing values:")
    print(df.isnull().sum())

    print("\nData types:")
    print(df.dtypes)
```

```python
# Convert date columns to datetime

order_items['shipping_limit_date'] = pd.to_datetime(
    order_items['shipping_limit_date'],
    errors='coerce'
)

orders_date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
    ]
for col in orders_date_columns:
    orders[col] = pd.to_datetime(orders[col], errors='coerce')

reviews['review_creation_date'] = pd.to_datetime(
    reviews['review_creation_date'],
    errors='coerce'
)

reviews['review_answer_timestamp'] = pd.to_datetime(
    reviews['review_answer_timestamp'],
    errors='coerce'
)

print("Date columns converted successfully!")
```

```python
# Verify datetime conversions

print("ORDER ITEMS")
print(order_items['shipping_limit_date'].dtype)
print("Missing:", order_items['shipping_limit_date'].isna().sum())

print('\nORDERS')
for col in orders_date_columns:
    print(col, "-", orders[col].dtype, "- Missing:", orders[col].isna().sum())

print('\nREVIEWS')
print("review_creation_date -", reviews['review_creation_date'].dtype,
      "- Missing:", reviews['review_creation_date'].isna().sum())

print("review_answer_timestamp -", reviews['review_answer_timestamp'].dtype,
      "- Missing:", reviews['review_answer_timestamp'].isna().sum())
```

```python
# Check order statuses and missing delivery dates

print("ORDER STATUS COUNTS")
print(orders['order_status'].value_counts())

print("\nMISSING CUSTOMER DELIVERY DATE BY STATUS")
print(
    orders[orders['order_delivered_customer_date'].isna()]
    ['order_status']
    .value_counts()
)

print("\nMISSING CARRIER DELIVERY DATE BY STATUS")
print(
    orders[orders['order_delivered_carrier_date'].isna()]
    ['order_status']
    .value_counts()
)

print("\nMISSING APPROVAL DATE BY STATUS")
print(
    orders[orders['order_approved_at'].isna()]
    ['order_status']
    .value_counts()
)
```

```python
# Clean geolocation data

# Remove duplicate rows
geolocation = geolocation.drop_duplicates().copy()

# Remove rows missing location information
geolocation = geolocation.dropna(
    subset=[
        'geolocation_lat',
        'geolocation_lng',
        'geolocation_city',
        'geolocation_state'
    ]
).copy()

# Verify Cleaning
print("Shape after cleaning:", geolocation.shape)
print("Duplicate rows:", geolocation.duplicated().sum())
print("\nMissing Values:")
print(geolocation.isnull().sum())
```

```python
# Inspect missing product data

print("TOTAL PRODUCTS:", len(products))

print("\nMISSING VALUES:")
print(products.isnull().sum())

print("\nPRODUCTS WITH MISSING CATEGORY:")
display(
    products[products['product_category_name'].isna()].head(10)
)

print("\nPRODUCTS WITH MISSING DIMENSIONS:")
display(
    products[
        products[
            ['product_weight_g',
             'product_length_cm',
             'product_height_cm',
             'product_width_cm']
        ].isna().any(axis=1)
    ]
)
```

```python
# Clean and translate product categories

# Fill missing original category names
products['product_category_name'] = (
    products['product_category_name'].fillna('unknown')
)

# Merge English category translations
products = products.merge(
category_translation,
on='product_category_name',
how='left'
)

# Fill categories that have no English translation
products['product_category_name_english'] = (
    products['product_category_name_english'].fillna('unknown')
)

# Verify
print(
    "Missing product categories:",
    products['product_category_name'].isna().sum()
)

print(
    "Missing English categories:",
    products['product_category_name_english'].isna().sum()
)

print("\nShape:", products.shape)

display(
    products[
        [
            'product_id',
            'product_category_name',
            'product_category_name_english']
        ].head(10)
)
```

```python
# Create main analysis dataset

analysis_df = order_items.merge(
    orders,
    on='order_id',
    how='left'
)

analysis_df = analysis_df.merge(
    products[
        ['product_id',
         'product_category_name',
         'product_category_name_english']
    ],
    on='product_id',
    how='left'
)

# Verify merge
print("Shape:", analysis_df.shape)

print("\nMissing values in key columns:")
print(
    analysis_df[
        ['order_id',
         'customer_id',
         'product_id',
         'price',
         'product_category_name_english']
    ].isnull().sum()
)

display(analysis_df.head())
```

```python
# Add customer information to analysis dataset

analysis_df = analysis_df.merge(
    customers[
        ['customer_id',
         'customer_unique_id',
         'customer_city',
         'customer_state']
    ],
    on='customer_id',
    how='left'
)

print("Shape:", analysis_df.shape)

print("\nMissing customer information:")
print(
    analysis_df[
        ['customer_id',
         'customer_unique_id',
         'customer_city',
         'customer_state']
    ].isnull().sum()
)
```

```python
# Add customer information

analysis_df.merge(
    customers[
        ['customer_id',
         'customer_unique_id',
         'customer_city',
         'customer_state']
    ],
    on='customer_id',
    how='left'
)

# Verify merge
print("Shape:", analysis_df.shape)

print("\nMissing customer information:")
print(
    analysis_df[
        ['customer_id',
         'customer_unique_id',
         'customer_city',
         'customer_state']
    ].isnull().sum()
)

print("\nUnique customers:",
      analysis_df['customer_unique_id'].nunique())

display(
    analysis_df[
        ['order_id',
         'customer_unique_id',
         'customer_city',
         'customer_state',
         'product_category_name_english',
         'price']
    ].head()
)
```

```python
# Add payment information

payment_summary = (
    payments.groupby('order_id')
    .agg(
        payment_value=('payment_value', 'sum'),
        payment_installments=('payment_installments', 'max'),
    )
    .reset_index()
)

analysis_df = analysis_df.merge(
    payment_summary,
    on='order_id',
    how='left'
)

# Verify
print("Shape:", analysis_df.shape)

print("\nMissing payment information:")
print(
    analysis_df[
        ['payment_value',
         'payment_installments']
    ].isnull().sum()
)

display(
    analysis_df[
        [
            'order_id',
            'customer_unique_id',
            'product_category_name_english',
            'price',
            'payment_value',
            'payment_installments']
    ].head()
)
```

```python
# Handle missing payment information

missing_payments = analysis_df[
    analysis_df['payment_value'].isna()
]

print("Rows with missing payment information:")
display(
    missing_payments[
        ['order_id',
         'customer_unique_id',
         'payment_value',
         'payment_installments']
    ]
)

# Remove rows without payment information
analysis_df = analysis_df.dropna(
    subset=['payment_value', 'payment_installments']
).copy()

# Verify
print("\nShape after removing missing payments:", analysis_df.shape)

print("\nRemaining missing payment values:")
print(
    analysis_df[
        ['payment_value',
         'payment_installments']
    ].isnull().sum()
)
```

```python
# Add reviews information

review_summary = (
    reviews.groupby('order_id')
    .agg(
        review_score=('review_score', 'mean')
    )
    .reset_index()
)

analysis_df = analysis_df.merge(
    review_summary,
    on='order_id',
    how='left'
)

# Verify
print("Shape:", analysis_df.shape)

print("\nMissing review scores:")
print(analysis_df['review_score'].isnull().sum())

print("\nReview score distribution:")
print(analysis_df['review_score'].value_counts().sort_index())

display(
    analysis_df[
        ['order_id',
         'customer_unique_id',
         'product_category_name_english',
         'price',
         'review_score']
    ].head()
)
```

```python
# Prepare order date information

date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_columns:
  analysis_df[col] = pd.to_datetime(
      analysis_df[col],
      errors='coerce'
  )

  # Create useful time-based columns
  analysis_df['purchase_year'] = analysis_df['order_purchase_timestamp'].dt.year
  analysis_df['purchase_month'] = analysis_df['order_purchase_timestamp'].dt.month
  analysis_df['purchase_month_name'] = analysis_df['order_purchase_timestamp'].dt.month_name()
  analysis_df['purchase_day'] = analysis_df['order_purchase_timestamp'].dt.day
  analysis_df['purchase_day_name'] = analysis_df['order_purchase_timestamp'].dt.day_name()

  # Calculate delivery time in days
  analysis_df['delivery_days'] = (
      analysis_df['order_delivered_customer_date'] -
      analysis_df['order_purchase_timestamp']
  ).dt.total_seconds() / 86400

  # Verify
  print("Shape:", analysis_df.shape)

  print("\nDate range:")
  print("First purchase:", analysis_df['order_purchase_timestamp'].min())
  print("Last purchase:", analysis_df['order_purchase_timestamp'].max())

  print("\nMissing delivery days:")
  print(analysis_df['delivery_days'].isnull().sum())

  display(
      analysis_df[
          ['order_id',
           'order_purchase_timestamp',
           'purchase_year',
           'purchase_month_name',
           'purchase_day_name',
           'delivery_days']
      ].head()
  )
```

```python
# Check orders with missing delivery dates

missing_delivery = analysis_df[
    analysis_df['delivery_days'].isna()
]

print("Rows with missing delivery days:", len(missing_delivery))

print("\nOrder status for missing delivery days:")
print(missing_delivery['order_status'].value_counts())
```

```python
# Check overall data quality before analysis

print("Dataset shape:", analysis_df.shape)

print("\nDuplicate rows:")
print(analysis_df.duplicated().sum())

print("\nMissing values:")
print(
    analysis_df[
        ['price',
         'payment_value',
         'review_score',
         'product_category_name_english',
         'delivery_days']
    ].isnull().sum()
)

print("\nNumeric summary:")
display(
    analysis_df[
        ['price',
         'payment_value',
         'review_score',
         'delivery_days']
    ].describe()
)
```

```python
# Analyse sales by product category

category_sales = (
    analysis_df.groupby('product_category_name_english')
    .agg(
        total_sales=('price', 'sum'),
        number_of_orders=('order_id', 'nunique'),
        average_price=('price', 'mean')
        )
    .sort_values('total_sales', ascending=False)
    .reset_index()
)

print("Top 10 product categories by total sales:")
display(category_sales.head(10))
```

```python
# Analyse monthly sales over time

monthly_sales = (
    analysis_df.groupby(['purchase_year', 'purchase_month'])
    .agg(
        total_sales=('price', 'sum'),
        number_of_orders=('order_id', 'nunique')
    )
    .reset_index()
    .sort_values(['purchase_year', 'purchase_month'])
)

monthly_sales['year_month'] = (
    monthly_sales['purchase_year'].astype(str) +
    '-' +
    monthly_sales['purchase_month'].astype(str).str.zfill(2)
)

print("Monthly sales:")
display(monthly_sales)
```

```python
# Analyse sales by day of the week

day_sales = (
    analysis_df.groupby('purchase_day_name')
    .agg(
        total_sales=('price', 'sum'),
        number_of_orders=('order_id', 'nunique'),
        average_order_value=('price', 'mean')
    )
    .reset_index()
)

day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

day_sales['purchase_day_name'] = pd.Categorical(
    day_sales['purchase_day_name'],
    categories=day_order,
    ordered=True
)

day_sales = day_sales.sort_values('purchase_day_name')

print("Sales by day of the week:")
display(day_sales)
```

```python
# Analyse review scores

review_analysis = (
    analysis_df.dropna(subset=['review_score'])
    .groupby('review_score')
    .agg(
        number_of_orders=('order_id', 'nunique'),
        average_price=('price', 'mean'),
        average_delivery_days=('delivery_days', 'mean')
    )
    .reset_index()
    .sort_values('review_score')
)

print("Review scores analysis:")
display(review_analysis)

print("\nOverall average review score:")
print(round(analysis_df['review_score'].mean(), 2))
```

```python
# Analyse delivery performance

delivery_analysis = (
    analysis_df.dropna(subset=['delivery_days'])
    .agg(
        average_delivery_days=('delivery_days', 'mean'),
        median_delivery_days=('delivery_days', 'median'),
        minimum_delivery_days=('delivery_days', 'min'),
        maximum_delivery_days=('delivery_days', 'max')
    )
)

print("Delivery performance:")
display(delivery_analysis)

# Compare delivery time with review score
delivery_reviews = (
    analysis_df.dropna(subset=['delivery_days', 'review_score'])
    .groupby('review_score')
    .agg(
        average_delivery_days=('delivery_days', 'mean'),
        number_of_orders=('order_id', 'nunique')
    )
    .reset_index()
)

print("\nDelivery time by review score:")
display(delivery_reviews)
```

```python
# Analyse payment installments

installment_analysis = (
    analysis_df.groupby('payment_installments')
    .agg(
        total_payment_value=('payment_value', 'sum'),
        number_of_orders=('order_id', 'nunique'),
        average_payment_value=('payment_value', 'mean')
    )
    .reset_index()
    .sort_values('number_of_orders', ascending=False)
)

print("Payment installment analysis:")
display(installment_analysis)

print("\nAverage number of installments:")
print(round(analysis_df['payment_installments'].mean(), 2))

print("\nMost common number of installments:")
print(analysis_df['payment_installments'].mode()[0])
```

```python
# Analyse sales by customer state

state_sales = (
    analysis_df.groupby('customer_state')
    .agg(
        total_sales=('price', 'sum'),
        number_of_orders=('order_id', 'nunique'),
        average_order_value=('price', 'mean')
    )
    .reset_index()
    .sort_values('total_sales', ascending=False)
)

print("Sales by customer state:")
display(state_sales)
```

```python
# Analyse freight costs

freight_analysis = (
    analysis_df.groupby('customer_state')
    .agg(
        total_freight_cost=('freight_value', 'sum'),
        average_freight=('freight_value', 'mean'),
        number_of_orders=('order_id', 'nunique')
    )
    .reset_index()
    .sort_values('total_freight_cost', ascending=False)
)

print("Freight cost analysis by customer state:")
display(freight_analysis)
```

```python
# Analyse average review score by delivery time

delivery_review = (
    analysis_df[['delivery_days', 'review_score']]
    .dropna()
    .copy()
)

# Group delivery times into ranges
delivery_review['delivery_group'] = pd.cut(
    delivery_review['delivery_days'],
    bins=[-1, 5, 10, 15, 20, 30, float('inf')],
    labels=[
        '0-5 days',
        '6-10 days',
        '11-15 days',
        '16-20 days',
        '21-30 days',
        '31+ days']
)

delivery_review_summary = (
    delivery_review.groupby('delivery_group', observed=False)
    .agg(
        average_review_score=('review_score', 'mean'),
        number_of_orders=('review_score', 'count')
    )
    .reset_index()
)

delivery_review_summary['average_review_score'] = (
    delivery_review_summary['average_review_score'].round(2)
)

print("Review scores by delivery time:")
display(delivery_review_summary)
```

```python
# Visualize average review score by delivery time

plt.figure(figsize=(12, 6))
plt.bar(
    delivery_review_summary['delivery_group'].astype(str),
    delivery_review_summary['average_review_score']
)

plt.xlabel('Delivery Time Group')
plt.ylabel('Average Review Score')
plt.title('Average Review Score by Delivery Time')
plt.ylim(0, 5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Measure the impact of longer delivery times on review scores

fast_delivery_score = delivery_review_summary.loc[
    delivery_review_summary['delivery_group'].astype(str) == '0-5 days',
    'average_review_score'
].iloc[0]

slow_delivery_score = delivery_review_summary.loc[
    delivery_review_summary['delivery_group'].astype(str) == '31+ days',
    'average_review_score'
].iloc[0]

score_drop = fast_delivery_score - slow_delivery_score
percent_drop = (score_drop / fast_delivery_score) * 100

print("Average review score for 0-5 day delivery:", fast_delivery_score)
print("Average review score for 31+ day delivery:", slow_delivery_score)
print("Drop in review score:", round(score_drop, 2))
print("Percentage decrease:", round(percent_drop, 2), "%")
```

```python
# Analyze review scores by product category

category_reviews = (
    analysis_df
    .groupby('product_category_name')
    .agg(
        average_review_score=('review_score', 'mean'),
        number_of_orders=('order_id', 'nunique')
    )
    .reset_index()
)

# Keep categories with at least 100 orders
category_reviews = category_reviews[
    category_reviews['number_of_orders'] >= 100
].copy()

category_reviews['average_review_score'] = (
    category_reviews['average_review_score'].round(2)
)

category_reviews = category_reviews.sort_values(
    'average_review_score',
    ascending=False
)

print("Top 10 product categories by average review score:")
display(category_reviews.head(10))
```
